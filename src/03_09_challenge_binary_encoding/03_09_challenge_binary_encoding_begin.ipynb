```python
from qiskit import *
from qiskit.visualization import plot_state_qsphere
%matplotlib inline
```

```python
### YOUR CODE GOES HERE ###
circuit = QuantumCircuit(4)
circuit.x([0,2,3]) # Apply X gate to qubits 0, 1, 2 and 3 to put it in superposition (configure 1101)
circuit.draw(output='mpl').show()
```

```python
# UPDATE: Code updated after recording to resolve breaking changes from Qiskit 1.0
circuit.remove_final_measurements()  # no measurements allowed
from qiskit.quantum_info import Statevector
statevector = Statevector(circuit)
plot_state_qsphere(statevector).show()
```
