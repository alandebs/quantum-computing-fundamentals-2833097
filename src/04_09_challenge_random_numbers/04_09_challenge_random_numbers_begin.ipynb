```python
from qiskit import *
from qiskit.visualization import plot_histogram
from math import pi
%matplotlib inline
```

```python
circuit = QuantumCircuit(3)

##### YOUR CODE GOES HERE #####
circuit.x(0)
circuit.h(1)
circuit.ry(pi/3, 2)
circuit.measure_all()
circuit.draw(output='mpl').show()
```

```python
# UPDATE: Code updated after recording to resolve breaking changes from Qiskit 1.0
from qiskit.providers.basic_provider import BasicProvider
simulator = BasicProvider().get_backend('basic_simulator')
compiled_circuit = transpile(circuit, simulator)
result = simulator.run(compiled_circuit, shots=1000000).result()
plot_histogram(result.get_counts()).show()
```
